```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("../data/raw/credit_default.xls",header=1)
target = "default payment next month"
display(df.head(4))
```

```python
df = df.drop(columns=["ID"])
```

```python
bill_cols = [
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6"
]
```

```python
payment_cols = [
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6"
]
```

```python
pay_status_cols = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]
```

```python
pay_status_cols = [
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]
```

```python
df["TOTAL_BILL_AMT"] = df[bill_cols].sum(axis=1)

df["TOTAL_PAY_AMT"] = df[payment_cols].sum(axis=1)

df["AVG_BILL_AMT"] = df[bill_cols].mean(axis=1)

df["AVG_PAY_AMT"] = df[payment_cols].mean(axis=1)

df["MAX_BILL_AMT"] = df[bill_cols].max(axis=1)

df["MAX_PAY_AMT"] = df[payment_cols].max(axis=1)

df["DELAY_COUNT"] = (df[pay_status_cols] > 0).sum(axis=1)

df["SEVERE_DELAY_COUNT"] = (df[pay_status_cols] >= 2).sum(axis=1)

df["AVG_PAY_STATUS"] = df[pay_status_cols].mean(axis=1)

df["MAX_PAY_STATUS"] = df[pay_status_cols].max(axis=1)
```

```python
df["PAYMENT_TO_BILL_RATIO"] = (
    df["TOTAL_PAY_AMT"] /
    df["TOTAL_BILL_AMT"].replace(0, np.nan)
)
```

```python
df["CREDIT_UTILIZATION"] = (
    df["BILL_AMT1"] /
    df["LIMIT_BAL"].replace(0, np.nan)
)
```

```python
df = df.replace([np.inf, -np.inf], np.nan)
```
